```python
import pandas as pd
df = pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

```python
df.info()
# Looking at the info its clear that the platform is quite new as the number of active customers is <4000
```

```python
df.drop(columns=['Customer ID']).describe(include = 'all')
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df.head()
```

```python
df['review_rating'] = df.groupby('category')['review_rating'].transform(lambda x: x.fillna(x.median()))
# Median obviously is a better idea than average because of possible outliers
# Also its unrealistic(not professional) to compare ratings of one genre with other
df.isnull().sum()
```

```python
# Introduce Age Groups
age_group = ['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = age_group)
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amt'})
df.columns
```

```python
df['frequency_of_purchases'].unique()
```

```python
# The customers working history is stored in string format - have to convert them to numbers
days = [14, 7, 365, 90, 14, 30, 90]
history = dict(zip(df['frequency_of_purchases'].unique(), days))
df['frequency_of_purchases'] = df['frequency_of_purchases'].map(history)
df = df.rename(columns = {'frequency_of_purchases' : 'purchase_frequency_days'})
df['purchase_frequency_days']
```

```python
# We dont need 'discount_applied' and 'promo_code_used' as one implies other - proof:
(df['discount_applied'] == df['promo_code_used']).all()
```
